```python
# ============================================================
# DAY 15 - MODEL EVALUATION
# STEP 1 - LOAD DATASET
# ============================================================

import re
import ast
import numpy as np
import pandas as pd

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    accuracy_score,
    confusion_matrix
)

# ------------------------------------------------------------
# LOAD YOUR CSV
# ------------------------------------------------------------

FILE_PATH = "/content/clean_jobs_descriptions_combined (1).csv"

df = pd.read_csv(FILE_PATH)

print("==========================================")
print("DATASET LOADED")
print("==========================================")

print("Dataset shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())


# ------------------------------------------------------------
# USE YOUR ACTUAL DATASET COLUMNS
# ------------------------------------------------------------

TEXT_COLUMN = "Job Description"
SKILL_COLUMN = "skills"


# ------------------------------------------------------------
# KEEP REQUIRED COLUMNS
# ------------------------------------------------------------

df = df[
    [TEXT_COLUMN, SKILL_COLUMN]
].copy()


# ------------------------------------------------------------
# HANDLE MISSING VALUES
# ------------------------------------------------------------

df[TEXT_COLUMN] = (
    df[TEXT_COLUMN]
    .fillna("")
    .astype(str)
)

df[SKILL_COLUMN] = (
    df[SKILL_COLUMN]
    .fillna("")
    .astype(str)
)


# ------------------------------------------------------------
# REMOVE EMPTY ROWS
# ------------------------------------------------------------

df = df[
    (df[TEXT_COLUMN].str.strip() != "") &
    (df[SKILL_COLUMN].str.strip() != "")
].reset_index(drop=True)


print("\nRows used:", len(df))


# ------------------------------------------------------------
# SHOW ACTUAL DATA
# ------------------------------------------------------------

print("\n==========================================")
print("SAMPLE DATA FROM YOUR CSV")
print("==========================================")

for i in range(min(5, len(df))):

    print("\nJob", i + 1)

    print("Job Description:")
    print(df.loc[i, TEXT_COLUMN][:500])

    print("\nSkills:")
    print(df.loc[i, SKILL_COLUMN])

    print("-" * 70)
```

```python
# ============================================================
# STEP 2 - CREATE CLEAN SKILL LIST
# ============================================================

COMMON_SKILLS = [
    "Python",
    "SQL",
    "Java",
    "JavaScript",
    "TypeScript",
    "HTML",
    "CSS",
    "C++",
    "C#",
    "Ruby",
    "PHP",

    "React",
    "Angular",
    "Vue",
    "Node.js",

    "Machine Learning",
    "Deep Learning",
    "Natural Language Processing",
    "NLP",

    "TensorFlow",
    "PyTorch",
    "Scikit-learn",

    "Pandas",
    "NumPy",
    "Matplotlib",

    "AWS",
    "Azure",
    "Google Cloud",
    "GCP",

    "Docker",
    "Kubernetes",

    "Git",
    "GitHub",
    "Linux",

    "Excel",
    "Power BI",
    "Tableau",

    "Spark",
    "Hadoop",

    "MongoDB",
    "MySQL",
    "PostgreSQL",
    "Oracle",

    "Django",
    "Flask",
    "FastAPI",

    "REST API",
    "API",

    "Data Analysis",
    "Data Visualization",
    "Statistics",
    "Data Science"
]


# ------------------------------------------------------------
# FUNCTION TO CHECK WHETHER SKILL EXISTS
# ------------------------------------------------------------

def contains_skill(text, skill):

    text = str(text).lower()
    skill = str(skill).lower()

    pattern = r"(?<![a-z0-9])" + re.escape(skill) + r"(?![a-z0-9])"

    return re.search(pattern, text) is not None


# ------------------------------------------------------------
# CREATE ACTUAL SKILLS
# ------------------------------------------------------------

def get_actual_skills(skill_text):

    found = []

    for skill in COMMON_SKILLS:

        if contains_skill(skill_text, skill):

            found.append(skill)

    return sorted(set(found))


df["actual_skills"] = df[SKILL_COLUMN].apply(
    get_actual_skills
)


# ------------------------------------------------------------
# REMOVE ROWS WITH NO RECOGNIZED SKILLS
# ------------------------------------------------------------

df = df[
    df["actual_skills"].apply(len) > 0
].reset_index(drop=True)


print("\n==========================================")
print("CLEAN SKILL DATA CREATED")
print("==========================================")

print("Rows remaining:", len(df))

print(
    "Number of skill classes:",
    len(COMMON_SKILLS)
)


# ------------------------------------------------------------
# SHOW ACTUAL SKILLS
# ------------------------------------------------------------

print("\nSample Actual Skills:")

for i in range(min(10, len(df))):

    print(
        f"Job {i + 1}:",
        df.loc[i, "actual_skills"]
    )
```

```python
# ============================================================
# STEP 3 - DICTIONARY MATCHING
# ============================================================

print("\n\n==========================================")
print("METHOD 1 - DICTIONARY MATCHING")
print("==========================================")


def dictionary_matching(text):

    predicted = []

    for skill in COMMON_SKILLS:

        if contains_skill(text, skill):

            predicted.append(skill)

    return sorted(set(predicted))


# ------------------------------------------------------------
# RUN DICTIONARY MATCHING
# ------------------------------------------------------------

df["Dictionary Matching"] = df[
    TEXT_COLUMN
].apply(dictionary_matching)


print("\nDictionary Matching completed!")


# ------------------------------------------------------------
# SHOW ACTUAL VS PREDICTED
# ------------------------------------------------------------

print("\n==========================================")
print("DICTIONARY MATCHING RESULTS")
print("==========================================")


for i in range(min(10, len(df))):

    print("\nJob", i + 1)

    print("Actual Skills    :")
    print(df.loc[i, "actual_skills"])

    print("Predicted Skills :")
    print(df.loc[i, "Dictionary Matching"])

    print("-" * 60)
```

```python
y_test = []
y_pred = []


for i in range(len(df)):

    actual = set(df.loc[i, "actual_skills"])
    predicted = set(df.loc[i, "Dictionary Matching"])

    for skill in COMMON_SKILLS:

        y_test.append(
            1 if skill in actual else 0
        )

        y_pred.append(
            1 if skill in predicted else 0
        )


dictionary_precision = precision_score(
    y_test,
    y_pred,
    zero_division=0
)

dictionary_recall = recall_score(
    y_test,
    y_pred,
    zero_division=0
)

dictionary_f1 = f1_score(
    y_test,
    y_pred,
    zero_division=0
)

dictionary_accuracy = accuracy_score(
    y_test,
    y_pred
)

dictionary_cm = confusion_matrix(
    y_test,
    y_pred
)


print("\n==========================================")
print("DICTIONARY MATCHING METRICS")
print("==========================================")

print(
    f"Precision : {dictionary_precision:.4f}"
)

print(
    f"Recall    : {dictionary_recall:.4f}"
)

print(
    f"F1 Score  : {dictionary_f1:.4f}"
)

print(
    f"Accuracy  : {dictionary_accuracy:.4f}"
)

print("\nConfusion Matrix:")
print(dictionary_cm)
```

```python
# ============================================================
# STEP 5 - REGEX METHOD
# ============================================================

print("\n\n==========================================")
print("METHOD 2 - REGEX")
print("==========================================")


def regex_matching(text):

    predicted = []

    for skill in COMMON_SKILLS:

        # Create regex pattern
        pattern = (
            r"(?<![a-zA-Z0-9])"
            + re.escape(skill)
            + r"(?![a-zA-Z0-9])"
        )

        if re.search(
            pattern,
            str(text),
            flags=re.IGNORECASE
        ):
            predicted.append(skill)

    return sorted(set(predicted))


# ------------------------------------------------------------
# RUN REGEX
# ------------------------------------------------------------

df["Regex"] = df[
    TEXT_COLUMN
].apply(regex_matching)


print("\nRegex completed!")


# ------------------------------------------------------------
# ACTUAL VS PREDICTED
# ------------------------------------------------------------

print("\n==========================================")
print("REGEX SAMPLE RESULTS")
print("==========================================")


for i in range(min(10, len(df))):

    print("\nJob", i + 1)

    print("Actual Skills    :")
    print(df.loc[i, "actual_skills"])

    print("Predicted Skills :")
    print(df.loc[i, "Regex"])

    print("-" * 60)
```

```python
# STEP 6 - REGEX METRICS

y_test_regex = []
y_pred_regex = []


for i in range(len(df)):

    actual = set(
        df.loc[i, "actual_skills"]
    )

    predicted = set(
        df.loc[i, "Regex"]
    )

    for skill in COMMON_SKILLS:

        y_test_regex.append(
            1 if skill in actual else 0
        )

        y_pred_regex.append(
            1 if skill in predicted else 0
        )


regex_precision = precision_score(
    y_test_regex,
    y_pred_regex,
    zero_division=0
)

regex_recall = recall_score(
    y_test_regex,
    y_pred_regex,
    zero_division=0
)

regex_f1 = f1_score(
    y_test_regex,
    y_pred_regex,
    zero_division=0
)

regex_accuracy = accuracy_score(
    y_test_regex,
    y_pred_regex
)

regex_cm = confusion_matrix(
    y_test_regex,
    y_pred_regex
)


print("\n==========================================")
print("REGEX METRICS")
print("==========================================")

print(f"Precision : {regex_precision:.4f}")
print(f"Recall    : {regex_recall:.4f}")
print(f"F1 Score  : {regex_f1:.4f}")
print(f"Accuracy  : {regex_accuracy:.4f}")

print("\nConfusion Matrix:")
print(regex_cm)
```

```python
# ============================================================
# STEP 7 - TF-IDF METHOD
# ============================================================

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


print("\n\n==========================================")
print("METHOD 3 - TF-IDF")
print("==========================================")


# ------------------------------------------------------------
# CREATE TF-IDF MODEL
# ------------------------------------------------------------

tfidf = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 3),
    min_df=1
)


# ------------------------------------------------------------
# FIT USING YOUR JOB DESCRIPTIONS + SKILLS
# ------------------------------------------------------------

documents = (
    df[TEXT_COLUMN].tolist()
    + COMMON_SKILLS
)


tfidf.fit(documents)


# ------------------------------------------------------------
# CREATE SKILL VECTORS
# ------------------------------------------------------------

skill_vectors = tfidf.transform(
    COMMON_SKILLS
)


# ------------------------------------------------------------
# TF-IDF PREDICTION FUNCTION
# ------------------------------------------------------------

def tfidf_matching(text, threshold=0.15):

    job_vector = tfidf.transform(
        [str(text)]
    )

    similarity_scores = cosine_similarity(
        job_vector,
        skill_vectors
    )[0]

    predicted = []

    for i, score in enumerate(
        similarity_scores
    ):

        if score >= threshold:

            predicted.append(
                COMMON_SKILLS[i]
            )

    return sorted(set(predicted))


# ------------------------------------------------------------
# RUN TF-IDF
# ------------------------------------------------------------

print("\nRunning TF-IDF...")

df["TF-IDF"] = df[
    TEXT_COLUMN
].apply(tfidf_matching)


print("TF-IDF completed!")


# ------------------------------------------------------------
# SHOW ACTUAL VS PREDICTED
# ------------------------------------------------------------

print("\n==========================================")
print("TF-IDF SAMPLE RESULTS")
print("==========================================")


for i in range(min(10, len(df))):

    print("\nJob", i + 1)

    print("Actual Skills    :")
    print(df.loc[i, "actual_skills"])

    print("Predicted Skills :")
    print(df.loc[i, "TF-IDF"])

    print("-" * 60)
```

```python
# ============================================================
# STEP 8 - TF-IDF METRICS
# ============================================================

y_true_tfidf = []
y_pred_tfidf = []


for i in range(len(df)):

    actual = set(
        df.loc[i, "actual_skills"]
    )

    predicted = set(
        df.loc[i, "TF-IDF"]
    )

    for skill in COMMON_SKILLS:

        y_true_tfidf.append(
            1 if skill in actual else 0
        )

        y_pred_tfidf.append(
            1 if skill in predicted else 0
        )


tfidf_precision = precision_score(
    y_true_tfidf,
    y_pred_tfidf,
    zero_division=0
)

tfidf_recall = recall_score(
    y_true_tfidf,
    y_pred_tfidf,
    zero_division=0
)

tfidf_f1 = f1_score(
    y_true_tfidf,
    y_pred_tfidf,
    zero_division=0
)

tfidf_accuracy = accuracy_score(
    y_true_tfidf,
    y_pred_tfidf
)

tfidf_cm = confusion_matrix(
    y_true_tfidf,
    y_pred_tfidf
)


print("\n==========================================")
print("TF-IDF METRICS")
print("==========================================")

print(f"Precision : {tfidf_precision:.4f}")
print(f"Recall    : {tfidf_recall:.4f}")
print(f"F1 Score  : {tfidf_f1:.4f}")
print(f"Accuracy  : {tfidf_accuracy:.4f}")

print("\nConfusion Matrix:")
print(tfidf_cm)
```

```python
# ============================================================
# STEP 9 - NER METHOD
# ============================================================

import spacy


print("\n\n==========================================")
print("METHOD 4 - NER")
print("==========================================")


# ------------------------------------------------------------
# CREATE BLANK NLP PIPELINE
# ------------------------------------------------------------

nlp = spacy.blank("en")


# ------------------------------------------------------------
# CREATE ENTITY RULER
# ------------------------------------------------------------

ruler = nlp.add_pipe(
    "entity_ruler"
)


# ------------------------------------------------------------
# CREATE SKILL PATTERNS
# ------------------------------------------------------------

patterns = []

for skill in COMMON_SKILLS:

    patterns.append({
        "label": "SKILL",
        "pattern": skill
    })


ruler.add_patterns(patterns)


# ------------------------------------------------------------
# NER PREDICTION
# ------------------------------------------------------------

def ner_matching(text):

    doc = nlp(str(text))

    predicted = []

    for ent in doc.ents:

        if ent.label_ == "SKILL":

            predicted.append(
                ent.text
            )

    return sorted(set(predicted))


# ------------------------------------------------------------
# RUN NER
# ------------------------------------------------------------

print("\nRunning NER...")

df["NER"] = df[
    TEXT_COLUMN
].apply(ner_matching)


print("NER completed!")


# ------------------------------------------------------------
# SHOW RESULTS
# ------------------------------------------------------------

print("\n==========================================")
print("NER SAMPLE RESULTS")
print("==========================================")


for i in range(min(10, len(df))):

    print("\nJob", i + 1)

    print("Actual Skills    :")
    print(df.loc[i, "actual_skills"])

    print("Predicted Skills :")
    print(df.loc[i, "NER"])

    print("-" * 60)
```

```python
# ============================================================
# STEP 10 - NER METRICS
# ============================================================

y_true_ner = []
y_pred_ner = []


for i in range(len(df)):

    actual = set(
        df.loc[i, "actual_skills"]
    )

    predicted = set(
        df.loc[i, "NER"]
    )

    for skill in COMMON_SKILLS:

        y_true_ner.append(
            1 if skill in actual else 0
        )

        y_pred_ner.append(
            1 if skill in predicted else 0
        )


ner_precision = precision_score(
    y_true_ner,
    y_pred_ner,
    zero_division=0
)

ner_recall = recall_score(
    y_true_ner,
    y_pred_ner,
    zero_division=0
)

ner_f1 = f1_score(
    y_true_ner,
    y_pred_ner,
    zero_division=0
)

ner_accuracy = accuracy_score(
    y_true_ner,
    y_pred_ner
)

ner_cm = confusion_matrix(
    y_true_ner,
    y_pred_ner
)


print("\n==========================================")
print("NER METRICS")
print("==========================================")

print(f"Precision : {ner_precision:.4f}")
print(f"Recall    : {ner_recall:.4f}")
print(f"F1 Score  : {ner_f1:.4f}")
print(f"Accuracy  : {ner_accuracy:.4f}")

print("\nConfusion Matrix:")
print(ner_cm)
```

```python
# ============================================================
# STEP 11 - MACHINE LEARNING MODEL
# Logistic Regression
# ============================================================

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MultiLabelBinarizer
from sklearn.multiclass import OneVsRestClassifier
from sklearn.linear_model import LogisticRegression


print("\n\n==========================================")
print("METHOD 5 - ML MODEL")
print("Algorithm: Logistic Regression")
print("==========================================")


# ------------------------------------------------------------
# MULTI-LABEL ENCODING
# ------------------------------------------------------------

mlb = MultiLabelBinarizer(
    classes=COMMON_SKILLS
)

Y = mlb.fit_transform(
    df["actual_skills"]
)


# ------------------------------------------------------------
# TRAIN TEST SPLIT
# ------------------------------------------------------------

X_train_text, X_test_text, y_train, y_test = train_test_split(

    df[TEXT_COLUMN],

    Y,

    test_size=0.20,

    random_state=42
)


print("\nTraining rows:", len(X_train_text))
print("Testing rows :", len(X_test_text))


# ------------------------------------------------------------
# TF-IDF FOR ML
# ------------------------------------------------------------

ml_tfidf = TfidfVectorizer(

    lowercase=True,

    stop_words="english",

    ngram_range=(1, 2),

    max_features=10000,

    min_df=2
)


X_train = ml_tfidf.fit_transform(
    X_train_text
)

X_test = ml_tfidf.transform(
    X_test_text
)


print("\nTraining TF-IDF shape:", X_train.shape)
print("Testing TF-IDF shape :", X_test.shape)


# ------------------------------------------------------------
# LOGISTIC REGRESSION
# ------------------------------------------------------------

ml_model = OneVsRestClassifier(

    LogisticRegression(
        max_iter=1000,
        class_weight="balanced"
    )
)


print("\nTraining Logistic Regression...")


ml_model.fit(
    X_train,
    y_train
)


print("Training completed!")


# ------------------------------------------------------------
# PREDICTION
# ------------------------------------------------------------

y_pred_ml = ml_model.predict(
    X_test
)


print("Prediction completed!")
```

```python
# ============================================================
# STEP 12 - ML MODEL METRICS
# ============================================================

ml_precision = precision_score(
    y_test,
    y_pred_ml,
    average="micro",
    zero_division=0
)

ml_recall = recall_score(
    y_test,
    y_pred_ml,
    average="micro",
    zero_division=0
)

ml_f1 = f1_score(
    y_test,
    y_pred_ml,
    average="micro",
    zero_division=0
)

ml_accuracy = accuracy_score(
    y_test,
    y_pred_ml
)


# ------------------------------------------------------------
# CONFUSION MATRIX
# ------------------------------------------------------------

ml_cm = confusion_matrix(
    y_test.ravel(),
    y_pred_ml.ravel()
)


print("\n==========================================")
print("LOGISTIC REGRESSION METRICS")
print("==========================================")

print(f"Precision : {ml_precision:.4f}")
print(f"Recall    : {ml_recall:.4f}")
print(f"F1 Score  : {ml_f1:.4f}")
print(f"Accuracy  : {ml_accuracy:.4f}")

print("\nConfusion Matrix:")
print(ml_cm)


# ------------------------------------------------------------
# ACTUAL VS PREDICTED
# ------------------------------------------------------------

print("\n==========================================")
print("LOGISTIC REGRESSION SAMPLE RESULTS")
print("==========================================")


test_indices = X_test_text.index


for row_number, original_index in enumerate(
    test_indices[:10]
):

    actual = [
        mlb.classes_[j]

        for j, value in enumerate(
            y_test[row_number]
        )

        if value == 1
    ]


    predicted = [
        mlb.classes_[j]

        for j, value in enumerate(
            y_pred_ml[row_number]
        )

        if value == 1
    ]


    print("\nJob", row_number + 1)

    print("Actual Skills    :")
    print(actual)

    print("Predicted Skills :")
    print(predicted)

    print("-" * 60)
```

```python
# STEP 13 - TRANSFORMER METHOD
# ============================================================

from transformers import pipeline

print("\n==========================================")
print("METHOD 6 - TRANSFORMER")
print("==========================================")

print("\nLoading Transformer model...")

transformer_classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli",
    device=-1   # CPU; use 0 if your Colab GPU is enabled
)

print("Transformer loaded!")


# ============================================================
# TRANSFORMER PREDICTION
# ============================================================

# Start with ONLY 5 rows for testing
TRANSFORMER_ROWS = min(5, len(df))

df_transformer = df.iloc[:TRANSFORMER_ROWS].copy()


# Use a smaller skill list first
TRANSFORMER_SKILLS = [
    "Python",
    "SQL",
    "AWS",
    "Machine Learning",
    "Power BI",
    "Excel",
    "Java",
    "JavaScript",
    "React",
    "Docker"
]


def transformer_matching(text):

    result = transformer_classifier(
        str(text),
        candidate_labels=TRANSFORMER_SKILLS,
        multi_label=True
    )

    predicted = []

    for label, score in zip(
        result["labels"],
        result["scores"]
    ):

        if score >= 0.50:
            predicted.append(label)

    return sorted(set(predicted))


print(
    f"\nRunning Transformer on {TRANSFORMER_ROWS} rows..."
)


df_transformer["Transformer"] = (
    df_transformer[TEXT_COLUMN]
    .apply(transformer_matching)
)


print("Transformer completed!")


# ============================================================
# SHOW RESULTS
# ============================================================

print("\n==========================================")
print("TRANSFORMER SAMPLE RESULTS")
print("==========================================")

for i in range(len(df_transformer)):

    print("\nJob", i + 1)

    print("Actual Skills    :")
    print(
        df_transformer.iloc[i]["actual_skills"]
    )

    print("Predicted Skills :")
    print(
        df_transformer.iloc[i]["Transformer"]
    )

    print("-" * 60)
```

```python

# STEP 14 - TRANSFORMER METRICS


y_true_transformer = []
y_pred_transformer = []


for i in range(
    len(df_transformer)
):

    actual = set(
        df_transformer.iloc[
            i
        ]["actual_skills"]
    )

    predicted = set(
        df_transformer.iloc[
            i
        ]["Transformer"]
    )


    for skill in COMMON_SKILLS:

        y_true_transformer.append(
            1 if skill in actual else 0
        )

        y_pred_transformer.append(
            1 if skill in predicted else 0
        )


transformer_precision = precision_score(
    y_true_transformer,
    y_pred_transformer,
    zero_division=0
)

transformer_recall = recall_score(
    y_true_transformer,
    y_pred_transformer,
    zero_division=0
)

transformer_f1 = f1_score(
    y_true_transformer,
    y_pred_transformer,
    zero_division=0
)

transformer_accuracy = accuracy_score(
    y_true_transformer,
    y_pred_transformer
)

transformer_cm = confusion_matrix(
    y_true_transformer,
    y_pred_transformer
)


print("\n==========================================")
print("TRANSFORMER METRICS")
print("==========================================")

print(
    f"Precision : {transformer_precision:.4f}"
)

print(
    f"Recall    : {transformer_recall:.4f}"
)

print(
    f"F1 Score  : {transformer_f1:.4f}"
)

print(
    f"Accuracy  : {transformer_accuracy:.4f}"
)

print("\nConfusion Matrix:")
print(transformer_cm)
```
