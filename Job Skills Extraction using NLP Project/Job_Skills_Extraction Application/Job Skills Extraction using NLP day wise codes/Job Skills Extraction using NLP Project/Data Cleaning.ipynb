```python
import pandas as pd
import re
```

```python
# Load dataset
df = pd.read_csv("/content/New Job Descriptions (1).csv")

# Check the data
df.head()
```

```python
print(df.columns)

print(df["Job Description"].head())
```

```python
def clean_text(text):

    # Convert to string
    text = str(text)

    # Remove HTML tags
    text = re.sub(r"<.*?>", " ", text)

    # Remove URLs
    text = re.sub(r"http\S+|www\S+", " ", text)

    # Remove email addresses
    text = re.sub(r"\S+@\S+", " ", text)

    # Remove phone numbers
    text = re.sub(r"\+?\d[\d\s\-()]{7,}\d", " ", text)

    # Remove special characters and symbols
    text = re.sub(r"[^a-zA-Z0-9\s]", " ", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text)

    # Convert to lowercase
    text = text.lower()

    # Remove leading and trailing spaces
    text = text.strip()

    return text
```

```python
df["clean_description"] = df["Job Description"].apply(clean_text)

df[["Job Description", "clean_description"]].head()
```

```python
print("Original Description:")
print(df["Job Description"].iloc[0])

print("\nCleaned Description:")
print(df["clean_description"].iloc[0])
```

```python
df.to_csv("clean_jobs_descriptions.csv", index=False)

print("Clean dataset saved successfully!")
```
