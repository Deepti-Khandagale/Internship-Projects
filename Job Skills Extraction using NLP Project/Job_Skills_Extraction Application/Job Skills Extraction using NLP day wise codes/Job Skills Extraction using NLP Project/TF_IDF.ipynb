```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
df=pd.read_csv("/content/clean_jobs_descriptions_combined (1).csv")
```

```python
vectorizer = TfidfVectorizer(
    max_features=1000,
    stop_words="english"
)

X = vectorizer.fit_transform(df["clean_description"].fillna(""))
```

```python
terms = vectorizer.get_feature_names_out()

print(terms[:20])
```

```python
#calculate the highest TF_IDF
import numpy as np

scores = X.mean(axis=0).A1

tfidf_df = pd.DataFrame({
    "term": terms,
    "tfidf_score": scores
})

tfidf_df = tfidf_df.sort_values(
    by="tfidf_score",
    ascending=False
)

tfidf_df.head(10)
```

```python
dictionary_skills = [
    "Python",
    "SQL",
    "Java",
    "JavaScript",
    "TypeScript",
    "Machine Learning",
    "Deep Learning",
    "Power BI",
    "Data Science",
    "Computer Vision"
]
```

```python
dictionary_skills_lower = {
    skill.lower() for skill in dictionary_skills
}
```

```python
tfidf_terms = set(tfidf_df["term"].str.lower())

matched_skills = [
    skill for skill in dictionary_skills
    if skill.lower() in tfidf_terms
]

print(matched_skills)
```

###Does TF-IDF identify useful skills automatically?
#

TF-IDF can identify important terms without a predefined skill dictionary.
It may discover useful technical terms such as python, sql, cloud, etc.
However, it doesn't understand whether a term is actually a skill.
Common words can receive high scores.
Multi-word skills can be split into separate words.
