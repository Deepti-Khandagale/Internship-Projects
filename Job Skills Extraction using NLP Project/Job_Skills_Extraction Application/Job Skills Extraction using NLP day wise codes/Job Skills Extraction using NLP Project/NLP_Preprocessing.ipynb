```python
#import libraries
import nltk
import spacy

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.stem import PorterStemmer
```

```python
#sample
text = """
We are looking for candidates with strong experience in Python and SQL.
Candidates should have experience working with Power BI and AWS.
They should also have good communication skills.
"""

print(text)
```

```python
#lowercase
lower_text = text.lower()

print(lower_text)
```

```python
tokens = word_tokenize(lower_text)

print(tokens)
```

```python
#stop_word removal
stop_words = set(stopwords.words('english'))

print(list(stop_words)[:20])
```

```python
filtered_tokens = []

for word in tokens:
    if word not in stop_words:
        filtered_tokens.append(word)

print(filtered_tokens)
```

```python
#remove punctuation
import string

clean_tokens = []

for word in filtered_tokens:
    if word not in string.punctuation:
        clean_tokens.append(word)

print(clean_tokens)
```

```python
#lemmatization
lemmatizer = WordNetLemmatizer()
```

```python
lemmatized_words = []

for word in clean_tokens:
    lemma = lemmatizer.lemmatize(word)
    lemmatized_words.append(lemma)

print(lemmatized_words)
```

```python
#stemming
stemmer = PorterStemmer()
```

```python
stemmed_words = []

for word in clean_tokens:
    stem = stemmer.stem(word)
    stemmed_words.append(stem)

print(stemmed_words)
```

```python
#sentence segmentation
sentences = sent_tokenize(text)

for sentence in sentences:
    print(sentence)
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")
```

```python
doc=nlp(text)
```

```python
#spacy tokenization
for token in doc:
    print(token.text)
```

```python
#spaCy stopword removal
spacy_filtered=[]
for token in doc:
  if not token.is_stop and not token.is_punct:
    spacy_filtered.append(token.text.lower())
print(spacy_filtered)
```

```python
#spacy Lemmatization
spacy_lemmas=[]
for token in doc:
  if not token.is_stop and not token.is_punct:
    spacy_lemmas.append(token.lemma_.lower())
print(spacy_lemmas)
```

```python
#preprocessing function
def preprocess_text(text):

    # Lowercase
    text = text.lower()

    # Tokenization
    tokens = word_tokenize(text)

    # Stopword removal
    stop_words = set(stopwords.words('english'))

    # Remove stopwords and punctuation
    clean_tokens = [
        word for word in tokens
        if word not in stop_words
        and word not in string.punctuation
    ]

    # Lemmatization
    lemmatizer = WordNetLemmatizer()

    lemmatized_tokens = [
        lemmatizer.lemmatize(word)
        for word in clean_tokens
    ]

    return lemmatized_tokens
```

```python
result = preprocess_text(text)

print(result)
```

```python
#converting the result to text
processed_text = " ".join(result)

print(processed_text)
```
