```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
import pandas as pd
```

```python
df=pd.read_csv("/content/clean_jobs_descriptions_combined (1).csv")
```

```python
vectorizer = TfidfVectorizer(
    ngram_range=(1, 3),
    stop_words="english",
)
```

```python
X = vectorizer.fit_transform(df["clean_description"].fillna(""))

terms = vectorizer.get_feature_names_out()
scores = X.mean(axis=0).A1

ngram_df = pd.DataFrame({
    "ngram": terms,
    "tfidf_score": scores
})

ngram_df = ngram_df.sort_values(by="tfidf_score", ascending=False)

ngram_df.head(10)
```

```python
unigrams = ngram_df[
    ~ngram_df["ngram"].str.contains(" ")
]

unigrams.head(20)
```

```python
bigrams = ngram_df[
    ngram_df["ngram"].str.count(" ") == 1
]

bigrams.head(20)
```

```python
trigrams = ngram_df[
    ngram_df["ngram"].str.count(" ") == 2
]

trigrams.head(20)
```
