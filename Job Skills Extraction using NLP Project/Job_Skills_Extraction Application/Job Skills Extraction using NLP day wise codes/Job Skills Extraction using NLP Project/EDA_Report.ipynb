```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("/content/New Job Descriptions (1).csv")
```

```python
#number of jobs
number_of_jobs = len(df)

print("Number of jobs:", number_of_jobs)
```

```python
#number of unique job titles
unique_job_titles = df["Job Title"].nunique()

print("Number of unique job titles:", unique_job_titles)
```

```python
#most common job roles
most_common_roles = df["Job Title"].value_counts()

print("Most Common Job Roles:")
display(most_common_roles.head(10))
```

```python
#jobs by location
jobs_by_location = df["location"].value_counts()

print("Jobs by Location:")
display(jobs_by_location)
```

```python
display(jobs_by_location.head(10))
```

```python
# Calculate description length
df["description_length"] = df["Job Description"].fillna("").str.len()

# Statistical summary
df["description_length"].describe()
```

```python
#average description length
average_description_length = df["description_length"].mean()

print(
    "Average description length:",
    round(average_description_length, 2)
)
```

```python
#minimum description length
minimum_description_length = df["description_length"].min()

print(
    "Minimum description length:",
    minimum_description_length
)
```

```python
#maximum description length
maximum_description_length = df["description_length"].max()

print(
    "Maximum description length:",
    maximum_description_length
)
```

```python
#missing values
missing_values = df.isnull().sum()

print("Missing Values:")
display(missing_values)
```

```python
duplicate_records = df.duplicated().sum()

print("Number of duplicate records:", duplicate_records)
```

#Visualizations

```python
#1.Top Job Titles
top_job_titles = df["Job Title"].value_counts().head(10)

plt.figure(figsize=(10, 6))
top_job_titles.plot(kind="bar")

plt.title("Top 10 Job Titles")
plt.xlabel("Job Title")
plt.ylabel("Number of Jobs")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
#2. Jobs by Location
jobs_by_location = df["location"].value_counts().head(10)

plt.figure(figsize=(10, 6))

jobs_by_location.plot(kind="bar")

plt.title("Top 10 Jobs by Locations")

plt.xlabel("Location")

plt.ylabel("Number of Jobs")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()

plt.show()
```

```python
#3.Description length Distribution
df["description_length"] = df["Job Description"].str.len()

plt.figure(figsize=(10, 6))

df["description_length"].plot(kind="hist", bins=30)

plt.title("Job Description Length Distribution")

plt.xlabel("Description Length")

plt.ylabel("Number of Jobs")

plt.tight_layout()

plt.show()
```

```python
#4.Missing values
missing = df.isnull().sum()
missing = missing[missing > 0]

if len(missing) == 0:
    print("No missing values found in the dataset.")
else:
    missing.plot(kind="bar")

    plt.title("Missing Values by Column")
    plt.xlabel("Column")
    plt.ylabel("Number of Missing Values")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
#5.Top Cmopanies
top_companies = df["Company"].value_counts().head(10)

plt.figure(figsize=(10, 6))

top_companies.plot(kind="bar")

plt.title("Top 10 Companies by Number of Jobs")

plt.xlabel("Company")

plt.ylabel("Number of Jobs")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()

plt.show()
```
