```python
import re
import pandas as pd
```

```python
df=pd.read_csv("/content/clean_jobs_descriptions_combined (1).csv")
```

```python
regex_patterns = {
    "Python": r"(?<!\w)(?:python3|Python|py)(?!\w)",
    "Java": r"(?<!\w)(?:java|Java)(?!\w)",
    "C++": r"(?<!\w)(?:C\+\+|c\+\+|cpp)(?!\w)",
    "JavaScript": r"(?<!\w)(?:javascript|JavaScript|js)(?!\w)",
    "HTML": r"(?<!\w)(?:\.html|html5|html|HTML)(?!\w)",
    "CSS": r"(?<!\w)(?:\.css|css3|CSS|css)(?!\w)",
    "SQL": r"(?<!\w)(?:sql|SQL)(?!\w)",
    "R": r"(?<!\w)(?:r\ programming|R)(?!\w)",
    "MySQL": r"(?<!\w)(?:MySQL|mysql)(?!\w)",
    "PostgreSQL": r"(?<!\w)(?:postgres\ sql|PostgreSQL|postgres)(?!\w)",
    "MongoDB": r"(?<!\w)(?:mongodb|MongoDB|mongo\ db)(?!\w)",
    "Oracle": r"(?<!\w)(?:oracle|Oracle|oracle\ database)(?!\w)",
    "Microsoft Excel": r"(?<!\w)(?:microsoft\ excel|MS\ Excel|Excel)(?!\w)",
    "Power BI": r"(?<!\w)(?:power\ bi|PowerBI|power-bi)(?!\w)",
    "Tableau": r"(?<!\w)(?:tableau|Tableau)(?!\w)",
    "AWS": r"(?<!\w)(?:aws|AWS|amazon\ web\ services)(?!\w)",
    "Azure": r"(?<!\w)(?:azure|Azure)(?!\w)",
    "Google Cloud": r"(?<!\w)(?:google\ cloud|Google\ Cloud|gcp|GCP)(?!\w)",
    "Data Science": r"(?<!\w)(?:data\ science|Data\ Science)(?!\w)",
    "Machine Learning": r"(?<!\w)(?:machine\ learning|Machine\ Learning|ml|ML)(?!\w)",
    "Deep Learning": r"(?<!\w)(?:deep\ learning|Deep\ Learning|dl|DL)(?!\w)",
    "Natural Language Processing": r"(?<!\w)(?:natural\ language\ processing|Natural\ Language\ Processing|nlp|NLP)(?!\w)",
    "Computer Vision": r"(?<!\w)(?:computer\ vision|Computer\ Vision|cv|CV)(?!\w)",
    "Git": r"(?<!\w)(?:git|Git)(?!\w)",
    "Docker": r"(?<!\w)(?:docker|Docker)(?!\w)",
    "Kubernetes": r"(?<!\w)(?:kubernetes|Kubernetes|k8s)(?!\w)",
    "Agile": r"(?<!\w)(?:agile|Agile)(?!\w)",
    "Scrum": r"(?<!\w)(?:scrum|Scrum)(?!\w)",
    "Project Management": r"(?<!\w)(?:project\ management|Project\ Management)(?!\w)",
    "Communication": r"(?<!\w)(?:communication|Communication)(?!\w)",
    "Teamwork": r"(?<!\w)(?:teamwork|Teamwork|team\ work)(?!\w)",
    "Problem Solving": r"(?<!\w)(?:problem\ solving|Problem\ Solving)(?!\w)",
    "Critical Thinking": r"(?<!\w)(?:critical\ thinking|Critical\ Thinking)(?!\w)",
    "Leadership": r"(?<!\w)(?:leadership|Leadership)(?!\w)",
    "Social Media": r"(?<!\w)(?:social\ media|Social\ Media)(?!\w)",
    "Marketing": r"(?<!\w)(?:marketing|Marketing)(?!\w)"
}
```

```python
def extract_skills(text):
    extracted = []

    if not isinstance(text, str):
        return extracted

    for skill, pattern in regex_patterns.items(): # Changed to regex_patterns
        if re.search(pattern, text, re.IGNORECASE):
            extracted.append(skill)

    return extracted
```

```python
df["extracted_skills"] = df["clean_description"].apply(extract_skills)
```

```python
test_description = df.loc[0, 'clean_description']
extracted = extract_skills(test_description)
print(f"Description: {test_description}")
print(f"Extracted Skills: {extracted}")
```
