```python
import pandas as pd
```

```python
from transformers import pipeline
```

```python
ner_pipeline = pipeline(        #load a pretrained NER model
    "ner",
    model="dslim/bert-base-NER",
    aggregation_strategy="simple"
)
```

```python
text = "Experience with Python, SQL and AWS."
# Pretrained Transformer NER
results = ner_pipeline(text)

print("Pretrained Transformer NER:")

for entity in results:

    print(
        entity["word"],
        "→",
        entity["entity_group"],
        "| Confidence:",
        round(entity["score"], 3)
    )
```

```python
text = """
The candidate should have experience with Python,
SQL, AWS, Power BI and Machine Learning.
"""

results = ner_pipeline(text)

for entity in results:
    print(
        entity["word"],
        "→",
        entity["entity_group"],
        "| Score:",
        round(entity["score"], 3)
    )
```

```python
skill_list = [
    "Python",
    "SQL",
    "AWS",
    "Power BI",
    "Machine Learning",
    "Deep Learning",
    "TensorFlow",
    "Tableau",
    "MySQL",
    "Microsoft Azure"
]
```

```python
import re

def extract_skills(text, skill_list):

    found_skills = []

    for skill in skill_list:

        pattern = r"\b" + re.escape(skill) + r"\b"

        if re.search(
            pattern,
            text,
            flags=re.IGNORECASE
        ):
            found_skills.append(skill)

    return found_skills
```

```python
# Custom Skill Extraction

custom_skills = extract_skills(
    text,
    skill_list
)

print("\nCustom Skill Extraction:")

for skill in custom_skills:
    print(skill, "→ SKILL")
```
